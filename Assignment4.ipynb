# Visualization with Seaborn and Matplotlib

### Use the “Titanic” dataset from Seaborn and do the following:

```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
data=sns.load_dataset("titanic")
data.head(5)
```

### 1. Countplot: Passenger Count by Embarked Port
● Create a countplot to display the number of passengers who boarded from each port (embarked), separated by sex.

```python
# 1. Countplot: Passenger Count by Embarked Port
plt.figure()
sns.countplot(data=data, x='embarked', hue='sex')
plt.title('Passenger Count by Embarked Port')
plt.show()
```

### 2. Swarmplot: Age Distribution across Classes
● Create a swarmplot to visualize the distribution of age across passenger classes(pclass), categorized by sex.

```python
# 2. Swarmplot: Age Distribution across Classes
plt.figure()
sns.swarmplot(data=data, x='pclass', y='age', hue='sex', dodge=True)
plt.title('Age Distribution across Classes')
plt.show()
```

### 3. Pie Chart: Gender Distribution
● Create a pie chart using Matplotlib to show the percentage of male vs. female passengers. - Add percentage labels using autopct - Use explode for one slice to highlight it.


```python
# 3. Pie Chart: Gender Distribution
plt.figure()
gender_counts = data['sex'].value_counts()
plt.pie(
    gender_counts, labels=gender_counts.index, autopct='%1.1f%%', explode=[0.1, 0]
)
plt.title('Gender Distribution')
plt.show()
```

### 4. Heatmap: Correlation Matrix
● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch,and survived. (Use seaborn.heatmap with annotations)


```python
# 4. Heatmap: Correlation Matrix
plt.figure()
cols = ['age', 'fare', 'pclass', 'sibsp', 'parch', 'survived']
sns.heatmap(data[cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

### 5. Violin Plot: Fare by Class and Gender
● Create a violin plot to visualize fare distribution across classes, separated by gender.

```python
 # 5. Violin Plot: Fare by Class and Gender
plt.figure()
sns.violinplot(data=data, x='pclass', y='fare', hue='sex', split=True)
plt.title('Fare by Class and Gender')
plt.show()
```

### 6. Scatter Plot: Age vs Fare
● Create a scatter plot to explore the relationship between age and fare, colored by survival status.

● Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels, and legends where needed.

```python
# 6. Scatter Plot: Age vs Fare
plt.figure()
sns.scatterplot(data=data, x='age', y='fare', hue='survived')
plt.title('Age vs Fare')
plt.show()
```
